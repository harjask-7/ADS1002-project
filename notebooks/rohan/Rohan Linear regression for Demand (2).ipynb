```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Define rainfall bands (with No Rain)
bins = [-0.01, 0, 5, 10, 20]
labels = ["No Rain", "Light (0–5 mm)", "Moderate (5–10 mm)", "Heavy (10–20 mm)"]

master_df["Precip_Band"] = pd.cut(
    master_df["Precipitation (mm)"], bins=bins, labels=labels
)

# Calculate mean demand per band, normalized within each state
banded = (
    master_df.groupby(["State", "Precip_Band"])["Total Demand"]
    .mean()
    .reset_index()
)

# Normalize to each state's mean demand (so baseline = 100%)
state_means = banded.groupby("State")["Total Demand"].transform("mean")
banded["Relative Demand (%)"] = (banded["Total Demand"] / state_means) * 100

# Ensure categorical ordering for consistency
banded["Precip_Band"] = pd.Categorical(
    banded["Precip_Band"],
    categories=labels,
    ordered=True
)

# Plot with color by Precip_Band (legend instead of x-axis labels)
g = sns.catplot(
    data=banded,
    x="State",  # now only state on x-axis
    y="Relative Demand (%)",
    hue="Precip_Band",
    kind="bar",
    height=5,
    aspect=1.6,
    palette="Blues"
)

# Axis labels
g.set_axis_labels("State", "Relative Demand (%)")

# Add legend with clear title
g._legend.set_title("Rainfall Band")

# Adjust layout
plt.subplots_adjust(top=0.85)
g.fig.suptitle("Relative Electricity Demand by Rainfall Band Across States", fontsize=14)

plt.show()
```

```python
import numpy as np
import pandas as pd
import statsmodels.formula.api as smf

# Work on a copy; rename to clean column names for the formula
df = master_df.copy().rename(columns={
    "Total Demand": "Demand",
    "Precipitation (mm)": "Precipitation"
})

rows = []

for state in df["State"].unique():
    sub = df.loc[df["State"] == state, ["Demand", "Precipitation", "State"]].dropna(subset=["Demand", "Precipitation"])

    # we find the ols
    model = smf.ols("Demand ~ Precipitation", data=sub).fit()
    slope = model.params["Precipitation"]  

    rows.append({
        "State": state,
        "Intercept": model.params["Intercept"],
        "Slope(Precipitation)": slope,
        "R2": model.rsquared
    })

    print(f"{state}: Intercept={model.params['Intercept']:.2f}, "
          f"Slope={slope:.4f}, R²={model.rsquared:.4f}")

per_state_precip_ols = pd.DataFrame(rows).sort_values("State").reset_index(drop=True)
per_state_precip_ols
```

As rain increases, it has a weak associatino with slightly lower demand, which could inidicate cooler/wetter condtions reducucing demand. Demand coudl then be said to be a factor of heating not cold/

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
import numpy as np
import pandas as pd


df = master_df.copy().rename(columns={
    "Total Demand": "Demand",
    "Air Temp (C)": "Temperature",
    "Humidity (%)": "Humidity",
    "Wind Speed (km/h)": "WindSpeed", 
})


# Parsing datetimes and creating hour,month,year
df["Datetime"] = pd.to_datetime(df["Datetime"], errors="coerce")
df = df.dropna(subset=["Datetime", "Demand", "Temperature", "Humidity", "Precipitation (mm)", "State"]).sort_values("Datetime")

df["Hour"]  = df["Datetime"].dt.hour
df["Month"] = df["Datetime"].dt.month
df["Year"]  = df["Datetime"].dt.year

#Building Ml table
feature_cols = ["Temperature", "Humidity", "Precipitation (mm)", "WindSpeed", "Hour", "Month", "Year", "State"]
use_cols = ["Datetime", "Demand"] + feature_cols
df_ml = df.filter(items=use_cols).copy()   # no KeyError if some are missing

df_ml = pd.get_dummies(df_ml, columns=["State"], drop_first=True)  # one-hot State

#Splitting the data
df_ml = df_ml.sort_values("Datetime")
split_idx = int(len(df_ml) * 0.90)  #We only use the first 90% of data so we dont use future data
train, test = df_ml.iloc[:split_idx].copy(), df_ml.iloc[split_idx:].copy() 

X_train = train.drop(columns=["Datetime", "Demand"])
y_train = train["Demand"].values
X_test  = test.drop(columns=["Datetime", "Demand"])
y_test  = test["Demand"].values


rf = RandomForestRegressor(
    n_estimators=200,
    max_depth=15,
    random_state=42,
    n_jobs=-1
)
rf.fit(X_train, y_train)
y_pred = rf.predict(X_test)

train_pred = rf.predict(X_train)
test_pred  = y_pred  
#We calculate RMSE, R^2, and MAE. note: 
#RMSE, the measure which linear regression aims to minimise, 
#MAE, similarly an error metric we want to minimise
scores = pd.DataFrame({
    "R^2": {
        "train": r2_score(y_train, train_pred),
        "test":  r2_score(y_test,  test_pred),
    },
    "RMSE": {
        "train": mean_squared_error(y_train, train_pred, squared=False),
        "test":  mean_squared_error(y_test,  test_pred,  squared=False),
    },
    "MAE": {
        "train": mean_absolute_error(y_train, train_pred),
        "test":  mean_absolute_error(y_test,  test_pred),
    },
})

print(scores.round(2))

# --- 7) Feature importance (top 10) ---
feat_importance = pd.Series(rf.feature_importances_, index=X_train.columns).sort_values(ascending=False)
print("\nTop 10 features:\n", feat_importance.head(10))
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import Lasso
from sklearn.metrics import r2_score
from sklearn.model_selection import train_test_split

# match data names---
df = master_df.copy().rename(columns={
    "Air Temp (C)": "Temperature",
    "Humidity (%)": "Humidity",
    "Wind Speed (km/h)": "Wind Speed (km/h)",  # keep original label
    "Total Demand": "Demand",                  # <- add this if your column is 'Total Demand'
})

# Time fields
df["Datetime"] = pd.to_datetime(df["Datetime"], errors="coerce")
df = df.dropna(subset=["Datetime", "Temperature", "Humidity", "Precipitation (mm)", "Demand"]).sort_values("Datetime")
df["Hour"]  = df["Datetime"].dt.hour
df["Month"] = df["Datetime"].dt.month
df["Year"]  = df["Datetime"].dt.year

# Temperature polynomials
df["t2"] = df["Temperature"]**2
df["t3"] = df["Temperature"]**3

# --- 2) Select target ---
features = ["Temperature", "t2", "t3", "Precipitation (mm)", "Humidity", "Wind Speed (km/h)", "Hour", "Month", "Year"]
target   = "Demand"

# One-hot encode (no categoricals here, but harmless) and drop missing
df_ml = pd.get_dummies(df[features + [target]], drop_first=True).dropna()

X = df_ml.drop(columns=[target])
y = df_ml[target].values

tail_n = 100_000
X, y = X.tail(tail_n), y[-tail_n:]

# --- 3) Chronological split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, shuffle=False)

# Standardise
mu, sigma = X_train.mean(), X_train.std().replace(0, 1)
X_train_n = (X_train - mu) / sigma
X_test_n  = (X_test  - mu) / sigma

# --- 4) Lasso 
alfas = np.logspace(-2, 2, num=21)  # 0.01 ... 100
lasso_training_score = np.zeros_like(alfas, dtype=float)
lasso_rsquared       = np.zeros_like(alfas, dtype=float)

for idx, alfa in enumerate(alfas):
    lasso = Lasso(alpha=alfa, max_iter=10000, random_state=42)
    lasso.fit(X_train_n, y_train)
    lasso_training_score[idx] = lasso.score(X_train_n, y_train)
    preds = lasso.predict(X_test_n)
    lasso_rsquared[idx] = r2_score(y_test, preds)

# --- 5) Plot train/test R² vs alpha (log x-axis) ---
plt.figure(figsize=(7,4))
plt.scatter(alfas, lasso_training_score, label='Training score')
plt.scatter(alfas, lasso_rsquared,       label='Testing score')
plt.xscale('log'); plt.xlim(0.01, 100)
plt.title('Lasso regularisation')
plt.xlabel('alpha'); plt.legend(loc='best'); plt.tight_layout()
plt.show()

# --- 6) Report optimal alpha & scores ---
opt_alfa_pos = int(np.argmax(lasso_rsquared))
opt_alfa = alfas[opt_alfa_pos]
print(f'Optimal alpha is {opt_alfa:.3f}')
print(f'Training score with optimal alpha is {lasso_training_score[opt_alfa_pos]:.3f}')
print(f'Testing  score with optimal alpha is {lasso_rsquared[opt_alfa_pos]:.3f}')
```
