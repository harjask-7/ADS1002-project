```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import matplotlib.pyplot as plt

# read master data
master_df = pd.read_csv(
    r"C:\Users\rohan\Downloads\master_data (1).csv",
    skipinitialspace=True,
    low_memory=False,
    parse_dates=['Datetime']
)



state_counts = master_df['State'].value_counts()
print(state_counts)
```

```python

plt.figure(figsize=(8,6))
sample_df = master_df.sample(n=50000, random_state=42)  # adjust size

sns.regplot(
    data=sample_df,
    x="Precipitation (mm)",
    y="Total Demand",
    
    scatter_kws={"alpha":0.3, "s":15}, 
    line_kws={"color":"red"}
)
plt.xlim(0,20)

plt.xlabel("Precipitation (mm)")
plt.ylabel("Total Demand (MW)")
plt.title("Relationship Between Precipitation and Electricity Demand Overall")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

## Overall Conclusion

Overall: There is a slight slope but alot of noise. When we look at states together there is no corelation at all. As rainfall increases demand generally decreases or doesnt change much, but never is higher. This suggests demand is due to air condtioning rather than heaters, as temp decreases as rainfall increases.

Vic: A negative slope, demand decreases as rainfall increases overall. This could suggest demand increases as temp increases and that demand is more due to air conditioning in heat rather than heaters in cold

NSW: Demand overall is a bit higher but has a more flat regression, so there isnt much link between rainfall and demand, or a very small link

Qld: The regression is almost flat, so we can say rainfall doesnt affect demand here. 

TAS: The demand here is much lower than all other places and has a flat slope with a slight downwards trend. Can conclude there isnt much affect of rainfall on demand

SA: Similar to SA, lower demand than other places and a weak negative slope. Rainfall doesnt affect demand much.
