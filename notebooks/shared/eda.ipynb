```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

master = pd.read_csv("/Users/jiehni/monash/ads1002/ADS1002-project/data/processed/master_data.csv")

master.shape
master.info()
master.describe()
```

```python
master.groupby("State")["Total Demand"].hist(bins=50, alpha=0.5, legend=True)
plt.show()
```

- Tasmania and South Australia are both very low demand states
    - Tasmania peaks very sharply at around 1000-1200 MW, may be due to it being a smaller state with lower populaton
- Queensland and Victoria are medium-high demand states
    - Reflects a larger population
- New South Wales has the highest demand for electricity

Note: would be beneficial to analyse the states separately 

```python
sns.heatmap(master.corr(numeric_only=True), annot=True, cmap="coolwarm")
```

Total Demand: 
- Strongest positive correlation with Air Temp (C)
- Strongest negative correlation with Wind Speed (km/h)
- Weak correlation with Humidity (%) and Precipitation (mm)
- Almost no correlation with RRP
